```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
df_task1 = pd.read_csv('Task1_Results.csv')

# Preprocessing: Ensure order_date is a datetime object for correct time-series plotting
df_task1['order_date'] = pd.to_datetime(df_task1['order_date'])

# Create the visualization
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_task1, x='order_date', y='new_customer_pct', marker='o', color='#E67E22', linewidth=2)

# Formatting according to instructions [cite: 40]
plt.title('Daily Percentage of New Customers Over Time', fontsize=14, fontweight='bold')
plt.xlabel('Order Date', fontsize=12)
plt.ylabel('New Customer Share (%)', fontsize=12)
plt.xticks(rotation=45)
plt.ylim(0, 105) # Percentages are best viewed on a 0-100 scale
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()

# Save the plot
plt.savefig('new_customer_line_chart.png')
plt.show()
```

### Observation
Daily new customer share trend reveals acquisition momentum and campaign effects.

### Explanation
This line chart plots new customer percentage by order date. A rising slope indicates that an increasing share of orders are from new customers, signaling successful customer acquisition. Sharp declines can indicate campaign lulls or retention focus changes.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
df_task2 = pd.read_csv('Task2_Results.csv')

# Preprocessing: Aggregate total orders by country and category
# Your Task2_Results.csv is at the SKU level, so we sum them for a category view
df_cat_country = df_task2.groupby(['country_iso', 'main_category'])['total_orders'].sum().reset_index()

# Create the visualization
plt.figure(figsize=(14, 7))
sns.barplot(data=df_cat_country, x='country_iso', y='total_orders', hue='main_category', palette='viridis')

# Formatting according to instructions [cite: 40]
plt.title('Total Order Volume by Product Category across Countries', fontsize=14, fontweight='bold')
plt.xlabel('Country (ISO Code)', fontsize=12)
plt.ylabel('Total Number of Orders', fontsize=12)
plt.legend(title='Main Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()

# Save the plot
plt.savefig('category_by_country_bar.png')
plt.show()
```

### Observation
The bar chart displays total order volume by product category across each country, with grouped bars indicating category composition per country.

### Explanation
This chart uses aggregated order counts per category-and-country, so it shows both overall volume and category mix by market. Taller bars mean higher total orders in that country, while different colored segments show which categories contribute most. It helps identify where certain categories are strong and where cross-country assortment or marketing should be prioritized.

Italy top 5 confirm the best-performing SKUs for inventory and promotion focus.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load your data
df = pd.read_csv('Task2_Results.csv')

# Filter for Italy Top 5 and sort
it_top = df[(df['country_iso'] == 'IT') & (df['popularity_label'] == 'Top 5')].sort_values('total_orders', ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(data=it_top, x='total_orders', y='sku', color='#27AE60') # Professional Green

plt.title('Italy: Top 5 Best Selling Products', fontsize=14, fontweight='bold')
plt.xlabel('Total Order Volume')
plt.ylabel('Product SKU')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('IT_Top_5.png')
plt.show()
```

### Observation
Italy top 5 confirm the best-performing SKUs for inventory and promotion focus.

### Explanation
This bar chart ranks the Top 5 Italian SKUs by total order volume. Higher bars represent stronger demand, so these SKUs are likely key revenue drivers in Italy. Use this insight to prioritize stock and marketing for these winning products.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Task 2 results
df = pd.read_csv('Task2_Results.csv')

# Filter for Italy and select the 5 least popular products
it_bottom_5 = df[df['country_iso'] == 'IT'].sort_values(['total_orders', 'sku']).head(5)

# Create the visualization
plt.figure(figsize=(10, 6))
sns.barplot(data=it_bottom_5, x='total_orders', y='sku', hue='sku', palette='magma')
plt.title('Top 5 Least Popular Products in Italy')
plt.xlabel('Total Orders')
plt.ylabel('Product SKU')
plt.legend([],[], frameon=False) # Removes redundant legend
plt.tight_layout()
plt.show()
```

### Observation
The bottom-5 Italy chart identifies low-demand products in the Italian market.

### Explanation
This bar plot ranks the five Italian SKUs with the lowest order volumes. The short bars show weak demand, signaling products that may be considered for discounting, seasonal rollout, or delisting. It helps operations decide where to reduce inventory exposure and shift focus to faster-moving SKUs.
